```python
# Dados dos clientes: pagamento em dia e alto engajamento
X = [[0, 0], [0, 1], [1, 0], [1, 1]]

# 1 libera a oferta, 0 não libera
y = [0, 0, 0, 1]

# Aqui vamos mostrar os registros, igual fizemos na aula
for i in range(len(X)):
    print("Cliente", i + 1)
    print("Pagamento em dia:", X[i][0])
    print("Alto engajamento:", X[i][1])
    print("Liberar oferta:", y[i])
    print("--------------------------")
```

```python
# Função degrau: transforma a soma em 0 ou 1
def degrau(valor):
    if valor >= 0:
        return 1
    else:
        return 0
```

```python
# Previsão sem bias: só usamos as entradas e os pesos
def prever_sem_vies(entrada, pesos):
    soma = 0
    i = 0
    while i < len(entrada):
        soma = soma + entrada[i] * pesos[i]
        i = i + 1
    return degrau(soma)

# Começamos com os pesos em zero, como pediu o professor
pesos_sem_vies = [0.0, 0.0]
taxa = 0.1
epocas = 10

for epoca in range(epocas):
    quantidade_erros = 0
    print("\n====================================")
    print("ÉPOCA SEM VIÉS", epoca + 1)
    print("====================================")

    for i in range(len(X)):
        previsto = prever_sem_vies(X[i], pesos_sem_vies)
        esperado = y[i]
        erro = esperado - previsto

        if erro != 0:
            quantidade_erros = quantidade_erros + 1

        print("\nCliente", i + 1)
        print("Entrada:", X[i])
        print("Esperado:", esperado)
        print("Previsto:", previsto)
        print("Erro:", erro)
        print("Pesos ANTES:", pesos_sem_vies[:])

        # Atualizamos cada peso usando o erro e a entrada
        for j in range(len(pesos_sem_vies)):
            pesos_sem_vies[j] = pesos_sem_vies[j] + taxa * erro * X[i][j]
            pesos_sem_vies[j] = round(pesos_sem_vies[j], 2)

        print("Pesos DEPOIS:", pesos_sem_vies[:])
        print("------------------------------------")

    print("RESUMO DA ÉPOCA", epoca + 1)
    print("Pesos finais:", pesos_sem_vies)
    print("Quantidade de erros:", quantidade_erros)

print("\nTREINAMENTO SEM VIÉS FINALIZADO")
print("Pesos finais:", pesos_sem_vies)
```

```python
# Agora a previsão tem bias, que ajuda a ajustar o limite da decisão
def prever(entrada, pesos, bias):
    soma = bias
    i = 0
    while i < len(entrada):
        soma = soma + entrada[i] * pesos[i]
        i = i + 1
    return degrau(soma)

# Os valores começam em zero, igual ao exercício da aula
pesos = [0.0, 0.0]
bias = 0.0
taxa = 0.1
epocas = 10

for epoca in range(epocas):
    quantidade_erros = 0
    print("\n====================================")
    print("ÉPOCA COM VIÉS", epoca + 1)
    print("====================================")

    for i in range(len(X)):
        previsto = prever(X[i], pesos, bias)
        esperado = y[i]
        erro = esperado - previsto

        if erro != 0:
            quantidade_erros = quantidade_erros + 1

        print("\nCliente", i + 1)
        print("Entrada:", X[i])
        print("Esperado:", esperado)
        print("Previsto:", previsto)
        print("Erro:", erro)
        print("Pesos ANTES:", pesos[:])
        print("Bias ANTES:", round(bias, 2))

        # Aqui fazemos a mesma atualização dos pesos que vimos em aula
        for j in range(len(pesos)):
            pesos[j] = pesos[j] + taxa * erro * X[i][j]
            pesos[j] = round(pesos[j], 2)

        # E agora atualizamos também o bias, porque ele também aprende
        bias = bias + taxa * erro
        bias = round(bias, 2)

        print("Pesos DEPOIS:", pesos[:])
        print("Bias DEPOIS:", bias)
        print("------------------------------------")

    print("RESUMO DA ÉPOCA", epoca + 1)
    print("Pesos finais da época:", pesos)
    print("Bias final da época:", bias)
    print("Quantidade de erros:", quantidade_erros)

    # Se não teve erros nesta época, podemos terminar o treino
    if quantidade_erros == 0:
        print("O modelo chegou a zero erros!")
        break

print("\nTREINAMENTO COM VIÉS FINALIZADO")
print("Épocas realizadas:", epoca + 1)
print("Pesos finais:", pesos)
print("Bias final:", bias)
```

```python
# Por último vamos testar os quatro clientes com o modelo treinado
for i in range(len(X)):
    previsto = prever(X[i], pesos, bias)
    print("Cliente", i + 1)
    print("Entrada:", X[i])
    print("Esperado:", y[i])
    print("Previsto:", previsto)
    if previsto == y[i]:
        print("Resultado: ACERTOU")
    else:
        print("Resultado: ERROU")
    print("--------------------------")
```
